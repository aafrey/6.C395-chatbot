```python
import pandas as pd
```

```python
# read in BPS data
bps_data = pd.read_csv("public_schools.csv")
bps_data.head()
```

```python
# Clean the data by:
# 1. Removing duplicates
# 2. Converting IDs into strings
# 3. Filling empty 'SHARED' values with 'Not Shared'
# 4. Filling empty 'COMPLEX' values with 'Not Complex'
# 5. Fixing Excel bug that turned some school types into dates 

# 1. Remove duplicates
print(f"Original number of rows: {len(bps_data)}")
bps_data = bps_data.drop_duplicates()
print(f"Number of rows after removing duplicates: {len(bps_data)}")

# 2. Convert IDs into strings
bps_data['SCH_ID'] = bps_data['SCH_ID'].astype(str)

# 3. Fill empty 'SHARED' values with 'Not Shared'
bps_data['SHARED'] = bps_data['SHARED'].fillna('Not Shared')

# 4. Fill empty 'COMPLEX' values with 'Not Complex'
bps_data['COMPLEX'] = bps_data['COMPLEX'].fillna('Not Part of Complex')
```

```python
# 5. Fix Excel bug that turned some school types into dates
print("Current unique values in 'SCH_TYPE':", bps_data['SCH_TYPE'].unique())
bps_data['SCH_TYPE'] = bps_data['SCH_TYPE'].replace({
    'ES': 'Elementary School',
    'MS': 'Middle School',
    'HS': 'High School',
    'K-8': 'K-8 School',
    'ELC': 'Early Learning Center',
    'Special': 'Special Education School',
    '6/7-12': '7-12 School', # This was likely misinterpreted as a date by Excel
})
print("Unique values in 'SCH_TYPE' after cleaning:", bps_data['SCH_TYPE'].unique())
```
